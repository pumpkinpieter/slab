# Speeding up saving 

Trying to speed up the animation saving process.  Testing here at first

```python
import numpy as np
import matplotlib.pyplot as plt
from slab import SlabExact
from IPython.display import HTML, Image
plt.style.use('dark_background')

%matplotlib widget
```

```python
# plt.rcParams["animation.html"] = "jshtml"
# plt.rcParams["animation.embed_limit"] = 120
```

```python
D = SlabExact(scale=1e-5, Ts=[6.5, 1, 8,1, 6.5],
              ns=[1, 1.44, 1, 1.44, 1], xrefs=[100, 40, 300, 40, 100],
              symmetric=True, wl=1e-6)

minwl, maxwl = 6.84e-6, 6.98e-6
fps = 32
secs = 8
Nwl = fps * secs
```

```python
ani = D.animate_propagation_constants(-.2, 2, -.7, .7,  filename='temp_test', colorbar=True,
                                wlmin=minwl, wlmax=maxwl, Nwl=Nwl, fps=fps, ext='mp4', figsize=(8, 5),
                                levels=60, rref=160, iref=160, color_min=-3, color_max=9, screen_dpi=266,
                               )
```

```python
width, height = 266*8/2, 266*5/2
width, height
```

# Conclusions

No clear way to speed up animation save time.  All methods in matplotlib.animation do end up saving the files to some form of file (BufferedWriter or Byte object), and the saving takes the same time regardless. However, the DPI does make a big difference here, so a smart way of choosing it so it looks good on the desired screen and at the desired size could yield significant improvements in time spent saving.

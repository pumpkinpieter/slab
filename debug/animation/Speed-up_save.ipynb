# Speeding up saving 

Trying to speed up the animation saving process.  Testing here at first

```python
import numpy as np
import matplotlib.pyplot as plt
from slab import SlabExact
from IPython.display import HTML, Image
plt.style.use('dark_background')

%matplotlib widget
```

```python
# plt.rcParams["animation.html"] = "jshtml"
# plt.rcParams["animation.embed_limit"] = 120
```

```python
D = SlabExact(scale=1e-5, Ts=[6.5, 1, 8,1, 6.5],
              ns=[1, 1.44, 1, 1.44, 1], xrefs=[100, 40, 300, 40, 100],
              symmetric=True, wl=1e-6)

minwl, maxwl = 6.84e-6, 6.98e-6
fps = 10
secs = 1
Nwl = fps * secs
```

```python
ani = D.animate_propagation_constants(-.2, 2, -.7, .7,  name='temp_test', colorbar=True,
                                wlmin=minwl, wlmax=maxwl, Nwl=Nwl, fps=fps, ext='mp4', figsize=(8, 5),
                                levels=60, rref=160, iref=160, color_min=-3, color_max=9, screen_dpi=266,
                                # no_save=True,
                               )
```

```python
width, height = 266*8/2, 266*5/2
width, height
```

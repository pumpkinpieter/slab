# Using ArtistAnimation instead of celluloid

We've been using the celluloid package and it does make things easier.  But it adds a dependency and an extra level of opacity (since I'm only partially comfortable with what it does under the hood).  Here I try to move to just using matplotlib animation classes along.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import ArtistAnimation
from slab import SlabExact
```

```python
from matplotlib import pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation 
 
# initializing a figure in 
# which the graph will be plotted
fig = plt.figure() 
 
# marking the x-axis and y-axis
axis = plt.axes(xlim =(0, 4), 
                ylim =(-2, 2)) 

# initializing a line variable
line, = axis.plot([], [], lw = 3) 
 
# data which the line will 
# contain (x, y)
def init(): 
    line.set_data([], [])
    return line,
 
def animate(i):
    x = np.linspace(0, 4, 1000)
 
    # plots a sine graph
    y = np.sin(2 * np.pi * (x - 0.01 * i))
    line.set_data(x, y)
    
    return line,
 
anim = FuncAnimation(fig, animate, init_func = init,
                     frames = 10, interval = 20, blit = True)

 
anim.save('continuousSineWave.mp4', 
          writer = 'ffmpeg', fps = 30)
```
